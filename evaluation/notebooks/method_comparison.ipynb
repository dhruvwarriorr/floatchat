# FloatChat-Lite anomaly method comparison

This notebook is a reproducible view over the frozen reviewed-case fixture. It deliberately refuses to invent labels: populate `evaluation/fixtures/anomaly_cases.csv` with independently reviewed cases and references before running the comparison. Generated numbers are not approved evidence until copied unchanged into the evidence log after review.

```python
from pathlib import Path
import json
import subprocess

ROOT = next(path for path in [Path.cwd(), *Path.cwd().parents] if (path / 'scripts' / 'evaluate_methods.py').is_file())
CASES = ROOT / 'evaluation' / 'fixtures' / 'anomaly_cases.csv'
OUTPUT = ROOT / 'evaluation' / 'results' / 'method_comparison.json'
print({'repository': str(ROOT), 'cases': str(CASES.relative_to(ROOT))})
```

```python
completed = subprocess.run([str(ROOT / '.venv' / 'bin' / 'python'), str(ROOT / 'scripts' / 'evaluate_methods.py')], cwd=ROOT, text=True, capture_output=True)
print(completed.stdout or completed.stderr)
if completed.returncode != 0:
    raise RuntimeError('Evaluation is correctly blocked until reviewed labels are supplied.')
```

```python
report = json.loads(OUTPUT.read_text())
report['methods']
```
